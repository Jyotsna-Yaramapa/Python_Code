"""Create a program to apply discount of 10% if the min amt is 600 and max cap of discount is 100 rupees.

Check for cart amt >= min cap amt else print the difference amount to add in the cart.
Further check for the disc percent applied cart amt is exceeding offer max amt, then apply flat offer max amt.
Else check for the disc percent applied cart amt is not exceeding offer max amt, then apply disc percentage"""

```python
from gettext import install

from prompt_toolkit.layout.processors import PasswordProcessor

min_cart_amount, max_cap_discount, disc_pct=600,100, 10
cat_value=float(input("Enter the cart amount:"))
if cat_value >= min_cart_amount:
    print("U you are eligible for offer")
    discount=cat_value*disc_pct/100
    print("Initial discount is :"+str(discount))
    if discount>=max_cap_discount:
       final_cart_value=cat_value-max_cap_discount
       discount=max_cap_discount
    else:
        final_cart_value=cat_value-discount
    print(f"{discount} is the discounted value")
    print("Final amount is:" + str(final_cart_value))
else:
    print("U you are not eligible for offer")
```
